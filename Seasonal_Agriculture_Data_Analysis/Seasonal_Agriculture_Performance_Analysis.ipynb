## 13. Outlier Analysis

Outlier analysis is performed to identify observations that differ substantially from the general distribution of numerical variables. The Interquartile Range (IQR) method is used to detect potential outliers.

```python
Q1 = df["Yield_Tonnes_Ha"].quantile(0.25)
Q3 = df["Yield_Tonnes_Ha"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
def count_crop_outliers(group):
    Q1 = group["Yield_Tonnes_Ha"].quantile(0.25)
    Q3 = group["Yield_Tonnes_Ha"].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return ((group["Yield_Tonnes_Ha"] < lower) |
            (group["Yield_Tonnes_Ha"] > upper)).sum()

crop_outlier_counts = df.groupby("Crop").apply(
    count_crop_outliers,
    include_groups=False
)

crop_outlier_counts.sort_values(ascending=False)
```

```python
def find_crop_outliers(group):
    Q1 = group["Yield_Tonnes_Ha"].quantile(0.25)
    Q3 = group["Yield_Tonnes_Ha"].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return group[
        (group["Yield_Tonnes_Ha"] < lower) |
        (group["Yield_Tonnes_Ha"] > upper)
    ]

crop_outliers = df.groupby("Crop", group_keys=False).apply(
    find_crop_outliers,
    include_groups=False
)

crop_outliers
```

## 20. Key Findings and Insights

The analysis of seasonal agricultural performance produced the following key insights:

1. **Kharif shows the strongest overall seasonal performance.**  
   Kharif has the highest average yield of 5.63 tonnes/ha, average production of 46.31 tonnes, and average profit of approximately ₹178,915.

2. **Zaid shows comparatively weaker economic performance.**  
   Zaid has the lowest average yield of 4.64 tonnes/ha and production of 38.89 tonnes. Its average profit is negative at approximately -₹24,805.

3. **Water usage differs across seasons.**  
   Zaid has the highest average water usage at approximately 6,420 m³, while Rabi has the lowest at approximately 5,847 m³.

4. **Water efficiency is highest in Kharif.**  
   Kharif has an average water efficiency of 5.89 tonnes per 1,000 m³, compared with 5.19 in Rabi and 4.41 in Zaid.

5. **Yield has a moderate positive association with profit.**  
   The correlation between yield and profit is approximately 0.49, indicating that farms with higher yields tend to have higher profits, although other factors also influence profitability.

6. **Water usage has a moderate positive association with yield.**  
   The correlation between water usage and yield is approximately 0.39. This indicates that higher water usage is associated with higher yield, but water usage alone does not explain yield differences.

7. **Irrigation methods show substantial differences in water efficiency.**  
   Rainfed farming has the highest average water efficiency at 7.56 tonnes per 1,000 m³, followed by Drip at 6.27, Sprinkler at 4.67, and Flood irrigation at 3.44.

8. **Agricultural performance varies considerably across states.**  
   Punjab records the highest average yield at 6.12 tonnes/ha, highest average production at 52.23 tonnes, and highest average profit at approximately ₹136,026. Andhra Pradesh records the lowest average yield, production, and profit among the states analyzed.

## 21. Evidence-Based Recommendations

Based on the findings from the analysis, the following recommendations can be made:

1. **Focus on improving Zaid-season performance.**  
   Zaid showed the lowest average yield, lowest water efficiency, and negative average profit. Farmers and planners should investigate the factors contributing to this weaker performance and identify suitable crop and resource-management strategies.

2. **Promote efficient water-management practices.**  
   The analysis shows substantial differences in water efficiency across irrigation methods. Water-efficient practices should be evaluated and promoted where they are suitable for local crops and environmental conditions.

3. **Reduce dependence on inefficient irrigation practices.**  
   Flood irrigation showed the lowest average water efficiency in the dataset. Opportunities to improve irrigation efficiency should be investigated, particularly in areas where water availability is limited.

4. **Consider profitability when selecting crops.**  
   Crop profitability varies considerably. Crop planning should consider expected yield, market price, production cost, and resource requirements rather than focusing on yield alone.

5. **Use state-specific agricultural strategies.**  
   Agricultural performance differs across states. State-level planning should consider local environmental conditions, crop composition, irrigation practices, and economic factors instead of applying the same strategy everywhere.

6. **Use multiple factors when planning agricultural production.**  
   The correlation analysis shows that agricultural outcomes are associated with several interconnected variables. Therefore, decisions should consider environmental conditions, resource usage, crop type, production, costs, and market conditions together.

7. **Monitor water efficiency alongside yield.**  
   Higher yield does not necessarily mean better resource efficiency. Water efficiency should be monitored alongside productivity to support sustainable agricultural planning.

8. **Investigate unusual observations instead of automatically removing them.**  
   Outlier analysis showed that some extreme yield values were related to crop-specific characteristics. Potential outliers should therefore be investigated for data quality and agricultural validity before being removed.

## 22. Limitations

6. The recommendations are based on patterns observed in the dataset and should be validated with local agricultural knowledge and additional data before being implemented.
